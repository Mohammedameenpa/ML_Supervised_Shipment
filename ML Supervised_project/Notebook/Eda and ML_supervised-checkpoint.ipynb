## 3. Outlier Visualization (Box Plots)

Outliers can heavily skew machine learning models. We use **Box Plots** (whisker plots) on our numerical features to visually isolate extreme data points that fall far outside the normal interquartile range (IQR).

````python
# Grab the first 4 numerical columns dynamically
num_cols = df.select_dtypes(include=[np.number]).columns[1:7]

if len(num_cols) > 0:
    fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(14, 10))
    axes = axes.flatten()

    for i, col in enumerate(num_cols):
        # Creating horizontal boxplots to cleanly view outlier points
        sns.boxplot(x=df[col], ax=axes[i], color='lightcoral', flierprops={"marker": "x", "markeredgecolor": "red"})
        axes[i].set_title(f'Outlier Detection Box Plot: {col}', fontsize=12)
        axes[i].set_xlabel('')

    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns available to check for outliers.")
````

## Outlier Detection via Interquartile Range (IQR) Method

This process identifies and counts the number of outliers present across all numerical features in the dataset using the statistical **Interquartile Range (IQR)** technique.



````python
# Select all numerical columns
num_cols = df.select_dtypes(include=[np.number]).columns

print("--- Outlier Count Per Column (IQR Method) ---")
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # Define bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Count how many rows fall outside the bounds
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"{col}: Found {len(outliers)} outliers")
````

## 4. Correlation Analysis (Heatmap)

We compute a Pearson correlation matrix and map it into a color-coded **Heatmap**. This allows us to spot linear dependencies between variables and see which features move together.

````python
numeric_df = df.select_dtypes(include=[np.number])

if numeric_df.shape[1] > 1:
    corr_matrix = numeric_df.corr()
    
    # Generate a mask to hide the top triangle since it mirrors the bottom half
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr_matrix, mask=mask, annot=True, fmt=".2f", cmap='Spectral', square=True, linewidths=0.5)
    plt.title("Numerical Feature Correlation Heatmap", fontsize=14, pad=15)
    plt.show()
else:
    print("Not enough numerical columns to calculate a correlation matrix.")
````

- Prevents data leakage by evaluating the model on unseen data.
- Provides a reliable estimate of the model's real-world performance.

## Analyzing Feature Importance

This cell uses **Permutation Importance** to measure how much each feature contributes to the model's predictions.

### Steps Performed

1. Calculate permutation importance using the best-performing model.
2. Randomly shuffle each feature and observe the impact on the model's F1-score.
3. Retrieve the names of the transformed features from the preprocessing pipeline.
4. Create a DataFrame containing feature names and their importance scores.
5. Sort the features in descending order of importance.

### Purpose

- Identify the most influential features in the model.
- Understand which variables have the greatest impact on predictions.
- Detect features with little or no contribution to model performance.
- Improve model interpretability and support feature selection.

### Interpretation

- **Higher importance score** → Feature has a strong influence on predictions.
- **Importance score near zero** → Feature has little impact on predictions.
- **Negative importance score** → Feature may be introducing noise and could potentially be removed.

````python
from sklearn.inspection import permutation_importance
import pandas as pd

result = permutation_importance(
    best_model,
    X_test_processed,
    y_test,
    scoring="f1",
    n_repeats=10,
    random_state=42
)

feature_names = preprocessor.get_feature_names_out()

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": result.importances_mean
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(importance_df)
````

````python
low_importance = importance_df[
    importance_df["Importance"] <= 0.001
]

print(low_importance)
````
